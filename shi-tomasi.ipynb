```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
border = cv2.imread('border.jpg')
gray = cv2.cvtColor(border, cv2.COLOR_BGR2GRAY)
```

```python
corners = cv2.goodFeaturesToTrack(
    gray,
    maxCorners=100,     
    qualityLevel=0.01,  
    minDistance=10      
)
```

```python
corners = np.intp(corners)
corners[:5]
```

```python
for corner in corners:
    x, y = corner.ravel()
    cv2.circle(gray, (x, y), 4, (0, 255, 0), -1)
```

```python
plt.imshow(gray,cmap="gray")
plt.axis("off")
```
